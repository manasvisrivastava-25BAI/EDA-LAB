```python
df = pd.DataFrame({
    "Age" : [25, np.nan, 30, 45, 50, 35, 40, 55],
    "Gender" : ['M', 'F', 'F', np.nan, 'M', 'M', 'F', 'F'],
    "Income" : [20000, 35000, 50000, 42000, 38000, 55000, np.nan, 60000]
})
df
```

```python
mean_imputer = SimpleImputer(strategy='mean')
df["Age_mean"] = mean_imputer.fit_transform(df[["Age"]])
df
```

```python
median_imputer = SimpleImputer(strategy="median")
df["Income_Median"] = median_imputer.fit_transform(df[["Income"]])
df
```
